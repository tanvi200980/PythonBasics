MAKE A DATAFRAME

```python
import pandas as pd
data={
    "Costumers":["A1","A2","A3","A4","A5"],
    "Age":[22,35,28,40,55],
    "City":["Delhi","Mathura","Lucknow","Vanarasi","Agra"],
    "Amount":[1200,3500,7800,4200,9500],
    "Rating":[4.5,3.5,2,4,5]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.loc[0:3,["Costumers","City","Amount"]]
```

```python
df.loc[df["Amount"]>5000,["Costumers","Amount"]]
```

```python
#first five rows and first 3 columns
df.iloc[0:5,0:3]
```
