```python
def add(*num):
  total=0
  for i in num:
    total+=i
  return total
```

```python
print(add(10,20))
print(add(10,20,30))
```

```python
def student(**details):
  print(details)

student(name="Rahul",age=20,city="Agra")
```

Reshaping Arrays

```python
import numpy as np
data=np.arange(1,13)
print("original data shape",data.shape)
a=data.reshape(3,4)
print("3x4 \n",a)
c=data.reshape(2,-1)
print("2xinfer \n",c)
```

Flatten() and ravel()

```python
data=np.array([[1,2,3],[4,5,6]])
print(data.ndim)
flat1=data.flatten()
print("flat1 shape:",flat1.shape)
flat2=data.ravel()
print("flat2 shape:",flat2.shape)
```

```python
marks=np.array([20,40,50,60,70,80,90])
passed= marks>=40
between=marks[(marks>=50) & (marks<=80)]
print(f'mask for marks>=40:{passed}')
print(f'Passing marks:{marks[passed]}')
print(f'Marks between 50 and 80:{between}')
print(passed)
print(between)
```
